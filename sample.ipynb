```python
# Attention Mechanism Simulation
# Shows how AI pays attention to important words

print("=" * 60)
print("🤖 CODE 2: Attention Mechanism Demo")
print("=" * 60)

# Simulate which words are "important" to each word
attention_scores = {
    "The": ["bank", "vault", "door"],  # "The" pays attention to these
    "bank": ["vault", "door"],          # "bank" relates to these
    "vault": ["door", "heavy"],         # "vault" relates to these
    "door": ["bank", "vault", "open"],  # "door" relates to these
    "was": ["heavy", "door"],           # "was" describes these
    "heavy": ["door", "open"],          # "heavy" affects these
    "I": ["open", "door"],              # "I" acts on these
    "couldn't": ["open"],               # "couldn't" affects this
    "open": ["door", "it"],             # "open" relates to these
    "it": ["door", "vault"]             # "it" refers to these
}

sentence = "The bank vault door was heavy. I couldn't open it."

print(f"\n📖 Sentence: {sentence}")
print("\n" + "=" * 60)
print("🎯 ATTENTION MECHANISM AT WORK:")
print("=" * 60 + "\n")

for word, attends_to in attention_scores.items():
    print(f"📌 Word: '{word}'")
    print(f"   ↳ Pays attention to: {', '.join(attends_to)}")

    # Show the connection
    if word == "it":
        print(f"   ✨ INSIGHT: '{word}' refers to 'door' (from 'vault door')")
    elif word == "bank":
        print(f"   ✨ INSIGHT: 'bank' = financial (not river) due to 'vault door'")
    print()

print("=" * 60)
print("💡 KEY INSIGHT:")
print("=" * 60)
print("""
Instead of reading word-by-word sequentially,
ATTENTION allows AI to see ALL connections at once.

This is why Transformers (2017) revolutionized AI!
They made attention the core mechanism.
""")
```
